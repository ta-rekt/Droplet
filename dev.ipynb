## Visualizing a 3D lattice

```python
# Generate a random 3D binary array
shape = (64, 64, 64)
arr = np.random.randint(0, 64**3, size=shape)

# Get coordinates where array == 1
coords = np.argwhere(arr <= 100)
x, y, z = coords[:, 0], coords[:, 1], coords[:, 2]

# Scatterplot with circles at sites with a 1
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(x, y, z, s=10, c='blue', marker='o', edgecolors='k', alpha=0.7)
ax.set_title('Random 3D Binary Array')
ax.set_xlabel('')
ax.set_ylabel('')
ax.set_zlabel('')
ax.set_xticks([])
ax.set_yticks([])
ax.set_zticks([])
for spine in ax.spines.values():
    spine.set_visible(False)
plt.show()
```
