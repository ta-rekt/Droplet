```python
import numpy as np
import networkx as nx
import matplotlib as mpl
import matplotlib.pyplot as plt
import importlib
import json
import os
```

## Visualizing polymers in a 3D lattice

```python
# import json data
dirname = "20250721/testpoly"
lattice = json.load(open("data/"+dirname+".json"))
lattice = np.array(lattice)
```

```python
# Get coordinates where array == 1
coords1 = np.argwhere(lattice[..., -1] == 1)
coords2 = np.argwhere(lattice[..., -2] == 1)
x1, y1, z1 = coords1[:, 0], coords1[:, 1], coords1[:, 2]
x2, y2, z2 = coords2[:, 0], coords2[:, 1], coords2[:, 2]

# Scatterplot with circles at sites with a 1
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(x1, y1, z1, s=10, c='blue', marker='o', edgecolors='k', alpha=0.7)
ax.scatter(x2, y2, z2, s=30, c='red', marker='o', edgecolors='k', alpha=0.7)
ax.set_title('Random 3D Binary Array')
ax.set_xlabel('')
ax.set_ylabel('')
ax.set_zlabel('')
ax.set_xlim(0, 16)
ax.set_ylim(0, 16)
ax.set_xticklabels('')
ax.set_yticklabels('')
ax.set_zticklabels('')
plt.show()
```
